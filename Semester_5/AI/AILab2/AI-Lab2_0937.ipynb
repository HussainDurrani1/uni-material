```python
def selection_sort(items):
    for step in range(len(items)):
        start = step
        idx = step

        for location in range(step, len(items)):
            if items[location] < items[idx]:
                idx = location
        
        items[start], items[idx] = items[idx], items[start]

    return items
```

```python
nums = [64, 25, 12, 131, 22, 11, 1, 43, 52]
print('Original list: ', nums)
```

```python
print('Sorted list: ', selection_sort(nums))
```

## Ex 2.2: 

```python
def merge(left, right):
    sorted_list = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            sorted_list.append(left[i])
            i += 1
        else:
            sorted_list.append(right[j])
            j += 1

    sorted_list.extend(left[i:])
    sorted_list.extend(right[j:])
    
    return sorted_list
```

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left_half = merge_sort(arr[:mid])
    right_half = merge_sort(arr[mid:])

    return merge(left_half, right_half)
```

```python
nums = [38, 27, 43, 3, 9, 82, 10]

print(f"Original List: {nums}")
```

```python
sorted_result = merge_sort(nums)

print(f"Final Sorted List: {sorted_result}")
```

## Ex 2.3: 

```python
def print_step(algo, pass_num, current_list):
    print(f"  Pass {pass_num}: {current_list}")

# 1. Bubble Sort
def bubble_sort(arr):
    print("\n--- Bubble Sort ---")
    arr = list(arr)
    n = len(arr)
    print(f"  Initial: {arr}")
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        print_step('Bubble ', i + 1, arr)
        if not swapped:
            print("  (Early exit: array already sorted)")
            break
```

```python

# 2. Selection Sort
def selection_sort(arr):
    print("\n--- Selection Sort ---")
    arr = list(arr)
    n = len(arr)
    print(f"  Initial: {arr}")
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
        print_step('Selection ', i + 1, arr)
```

```python
# 3. Insertion Sort
def insertion_sort(arr):
    print("\n--- Insertion Sort ---")
    arr = list(arr)
    n = len(arr)
    print(f"  Initial: {arr}")
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
        print_step('Insertion ', i, arr)
```

```python
# 4. Merge Sort (Traces merges)
def merge_sort(arr, level=0, side="Root"):
    if len(arr) > 1:
        mid = len(arr) // 2
        left = arr[:mid]
        right = arr[mid:]

        merge_sort(left, level + 1, "Left")
        merge_sort(right, level + 1, "Right")

        i = j = k = 0
        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                arr[k] = left[i]
                i += 1
            else:
                arr[k] = right[j]
                j += 1
            k += 1

        while i < len(left):
            arr[k] = left[i]
            i += 1
            k += 1

        while j < len(right):
            arr[k] = right[j]
            j += 1
            k += 1
        
        print(f"  Merged sub-array: {arr}")
    return arr
```

```python
chars = ['P', 'Y', 'T', 'H', 'O', 'N']

bubble_sort(chars)
selection_sort(chars)
insertion_sort(chars)

print("\n--- Merge Sort ---")
print(f"  Initial: {chars}")
sorted_merge = merge_sort(list(chars))
```

## Ex 2.4: 

```python
def calculate_stats():
    user_input = input("Enter numbers separated by spaces: ")
    num_list = [float(x) for x in user_input.split()]
    
    if not num_list:
        print("The list is empty.")
        return

    total_sum = sum(num_list)
    n = len(num_list)

    mean = total_sum / n

    sorted_list = sorted(num_list)
    if n % 2 != 0:
        median = sorted_list[n // 2]
    else:
        median = (sorted_list[(n // 2) - 1] + sorted_list[n // 2]) / 2

    frequency_dict = {}
    for num in num_list:
        frequency_dict[num] = frequency_dict.get(num, 0) + 1

    max_count = max(frequency_dict.values())
    modes = [k for k, v in frequency_dict.items() if v == max_count]

    print(f"\n--- Results ---")
    print(f"List: {num_list}")
    print(f"Mean: {mean:.2f}")
    print(f"Median: {median}")
    print(f"Mode(s): {modes if max_count > 1 else 'No repeating mode'}")

calculate_stats()
```

## Ex 2.5:

```python
def dups(input_list):
    seen = set()
    duplicates = set()
    for item in input_list:
        if item in seen:
            duplicates.add(item)
        else:
            seen.add(item)
    return list(duplicates)

user_elements = input("Enter items separated by spaces to find duplicates: ").split()

result = dups(user_elements)

print(f"Duplicate elements: {result}")
```

## Ex 2.9: 

```python
import time

big_list = list(range(100000, 300001))

def search(integer_list, key):
    for i in range(0, len(integer_list)):
        if integer_list[i] == key:
            return i
    return None

def timer(key):
    start = time.time()
    search(big_list, key)
    end = time.time()
    return end - start

print(f"i.   Front (100000):     {timer(100000):.6f} seconds")
print(f"ii.  Middle (200000):    {timer(200000):.6f} seconds")
print(f"iii. Back (300000):      {timer(300000):.6f} seconds")
print(f"iv.  Not in list (3):    {timer(3):.6f} seconds")
```

```python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    return sum(1 for char in text if char in vowels)

print(count_vowels("Artificial"))  # Output: 5
```

## Ex: 3.3

```python
def is_palindrome(text):
    cleaned = text.lower()
    return cleaned == cleaned[::-1]

print(is_palindrome("madam"))  # Output: True
print(is_palindrome("Hello"))  # Output: False
```

## Ex: 3.11

```python
def reverse_string(text):
    # Using a for loop
    reversed_loop = ""
    for char in text:
        reversed_loop = char + reversed_loop
        
    # Using slicing
    reversed_slice = text[::-1]
    
    return reversed_loop, reversed_slice

loop_res, slice_res = reverse_string("python")
print(f"Loop result: {loop_res}")
print(f"Slice result: {slice_res}")
print(f"Match: {loop_res == slice_res}")
```

## Ex: 3.14

```python
import math

try:
    n = int(input("Enter a positive integer n: "))
    if n <= 0:
        print("Please enter a positive integer greater than 0.")
    else:
        # 1. Even numbers from 1 to n
        evens = [x for x in range(1, n + 1) if x % 2 == 0]
        print(f"Even numbers: {evens}")
        
        # 2. Sum of numbers from 1 to n
        total_sum = sum(range(1, n + 1))
        print(f"Sum from 1 to {n}: {total_sum}")
        
        # 3. Factorial using a while loop
        factorial_loop = 1
        i = 1
        while i <= n:
            factorial_loop *= i
            i += 1
            
        math_factorial = math.factorial(n)
        print(f"Factorial (while loop): {factorial_loop}")
        print(f"Factorial (math module): {math_factorial}")
except ValueError:
    print("Invalid integer entered.")
```

## Ex: 3.15

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

primes = [n for n in range(1, 101) if is_prime(n)]
print("Primes between 1 and 100:", primes)
```

## Ex: 3.18

```python
# 1. Create marks.txt with at least five students
student_records = [
    "Ali,85",
    "Sara,92",
    "Ahmed,58",
    "Zainab,74",
    "Bilal,45"
]

with open("marks.txt", "w") as f:
    for record in student_records:
        f.write(record + "\n")

# 2. Read file and store in dictionary
students = {}
with open("marks.txt", "r") as f:
    for line in f:
        if line.strip():
            name, mark = line.strip().split(",")
            students[name] = int(mark)

# 3. Print each student's name/mark, class average, highest student
print("Student Marks:")
for name, mark in students.items():
    print(f"- {name}: {mark}")

avg_mark = sum(students.values()) / len(students)
top_student = max(students, key=students.get)

print(f"\nClass Average: {avg_mark:.2f}")
print(f"Highest Score: {top_student} ({students[top_student]})")

# 4. Write below average students to below_average.txt
with open("below_average.txt", "w") as f:
    for name, mark in students.items():
        if mark < avg_mark:
            f.write(f"{name},{mark}\n")

print("Below average students written to below_average.txt")
```

## Ex: 3.19

```python
def check_password(password):
    if len(password) < 8:
        return False
    
    has_digit = False
    has_upper = False
    
    for char in password:
        if char.isdigit():
            has_digit = True
        if char.isupper():
            has_upper = True
            
    return has_digit and has_upper

while True:
    pwd = input("Enter a valid password (min 8 chars, 1 digit, 1 uppercase): ")
    if check_password(pwd):
        print("Password accepted successfully!")
        break
    else:
        print("Invalid password. Please try again.")
```

## Ex: 3.20

```python
import os

contacts = {}

# Load contacts from file on start if it exists
if os.path.exists("contacts.txt"):
    with open("contacts.txt", "r") as f:
        for line in f:
            if "," in line:
                name, phone = line.strip().split(",", 1)
                contacts[name] = phone

while True:
    print("\n--- Mini Contact Book ---")
    print("1. Add a contact")
    print("2. Search for a contact")
    print("3. Delete a contact")
    print("4. Show all contacts")
    print("5. Exit")
    
    choice = input("Choose an option (1-5): ")
    
    if choice == '1':
        name = input("Enter name: ").strip()
        phone = input("Enter phone number: ").strip()
        contacts[name] = phone
        print(f"Contact '{name}' added.")
    elif choice == '2':
        name = input("Enter name to search: ").strip()
        if name in contacts:
            print(f"{name}: {contacts[name]}")
        else:
            print("Contact not found.")
    elif choice == '3':
        name = input("Enter name to delete: ").strip()
        if name in contacts:
            del contacts[name]
            print(f"Contact '{name}' deleted.")
        else:
            print("Contact not found.")
    elif choice == '4':
        if contacts:
            print("\nSaved Contacts:")
            for name, phone in contacts.items():
                print(f"- {name}: {phone}")
        else:
            print("Contact book is empty.")
    elif choice == '5':
        # Save contacts to file on exit
        with open("contacts.txt", "w") as f:
            for name, phone in contacts.items():
                f.write(f"{name},{phone}\n")
        print("Contacts saved. Exiting application. Goodbye!")
        break
    else:
        print("Invalid option. Please choose between 1 and 5.")
```

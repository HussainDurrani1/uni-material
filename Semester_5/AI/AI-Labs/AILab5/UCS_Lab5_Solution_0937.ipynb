## Exercise 5.5 — UCS for All Nodes

```python
# Exercise 5.5 - UCS for All Nodes
def ucs_all(start):
    queue = [(0, start)]
    visited = set()
    costs = {}

    while queue:
        cost, current = heapq.heappop(queue)

        if current in visited:
            continue

        visited.add(current)
        costs[current] = cost

        for node, weight in graph[current].items():
            if node not in visited:
                heapq.heappush(queue, (cost + weight, node))

    return costs

graph = graph3

result = ucs_all('S')
print(result)
print("Farthest Node:", max(result, key=result.get))
# Written answer: The farthest node from S is G, with a
# minimum path cost of 9.
```

## Exercise 4.2:

```python
graph1 = {
    'S': ['A', 'B'],
    'A': ['C', 'D'],
    'B': ['E'],
    'C': ['F'],
    'D': ['G'],
    'E': [],
    'F': ['G'],
    'G': []
}

graph = graph1

def bfs(start, goal):
    queue = [start]
    visited = []

    while queue:
        node = queue.pop(0)

        if node not in visited:
            visited.append(node)

            if node == goal:
                return visited

            for neighbor in sorted(graph[node]):
                queue.append(neighbor)

    return None


def dfs(start, goal):
    stack = [start]
    visited = []

    while stack:
        node = stack.pop()

        if node not in visited:
            visited.append(node)

            if node == goal:
                return visited

            for neighbor in sorted(graph[node], reverse=True):
                stack.append(neighbor)

    return None


print("BFS:", bfs('S', 'G'))
print("DFS:", dfs('S', 'G'))
```

```python
graph = graph1

def bfs(start, goal):
    queue = [start]
    visited = set()
    parent = {start: None}

    while queue:
        node = queue.pop(0)

        if node == goal:
            path = []
            while node is not None:
                path.append(node)
                node = parent[node]
            return path[::-1]

        if node not in visited:
            visited.add(node)

            for neighbor in sorted(graph[node]):
                if neighbor not in parent:
                    parent[neighbor] = node
                    queue.append(neighbor)

    return None


print("Actual Path:", bfs('S', 'G'))
```
